```python
df['goals_per_90'] = df['goals_scored'] / df['minutes'] * 90
df['assists_per_90'] = df['assists'] / df['minutes'] * 90
df['goal_involvements_per_90'] = df['goals_per_90'] + df['assists_per_90']
```

```python
df['underperformance_index'] = df['expected_goal_involvements_per_90'] / df['goal_involvements_per_90']
```

```python
df.query('underperformance_index != inf').sort_values(['underperformance_index'], ascending = False).head(10)
```

```python
df['underperformance_and_form_index'] = df['underperformance_index'] + df['form']
```

```python
df['underperformance_and_form_and_ppg_index'] = df['underperformance_and_form_index'] + df['points_per_game']
```

```python
df_top_ofensivos = df.query('underperformance_index != inf and underperformance_index > 1').sort_values(['underperformance_and_form_and_ppg_index'], ascending = False)
```
